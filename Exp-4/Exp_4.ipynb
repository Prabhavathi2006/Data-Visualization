```python
# ============================================================
# LIFE EXPECTANCY DATA ANALYSIS
# ============================================================

# ------------------------------------------------------------
# 1. IMPORT LIBRARIES
# ------------------------------------------------------------

import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import MinMaxScaler, LabelEncoder


# ------------------------------------------------------------
# 2. LOAD DATASET
# ------------------------------------------------------------

df = pd.read_csv("Life Expectancy Data.csv")

# Remove extra spaces from column names
df.columns = df.columns.str.strip()

print(df.head())


# ------------------------------------------------------------
# 3. DATA UNDERSTANDING
# ------------------------------------------------------------

print("Dataset Shape :", df.shape)

print("\nDataset Information:")
df.info()

print("\nData Types:")
print(df.dtypes)


# ------------------------------------------------------------
# 4. STATISTICAL SUMMARY
# ------------------------------------------------------------

print("\nStatistical Summary:")
print(df.describe())


# ------------------------------------------------------------
# 5. MISSING VALUES
# ------------------------------------------------------------

print("\nMissing Values:")
print(df.isnull().sum())


# ------------------------------------------------------------
# 6. MISSING VALUE PERCENTAGE
# ------------------------------------------------------------

missing_percentage = (
    df.isnull().sum() / len(df)
) * 100

print("\nMissing Value Percentage:")
print(missing_percentage)


# ------------------------------------------------------------
# 7. DUPLICATE ROWS
# ------------------------------------------------------------

print(
    "\nDuplicate Rows :",
    df.duplicated().sum()
)

df.drop_duplicates(inplace=True)

print(
    "After Duplicate Removal :",
    df.shape
)


# ------------------------------------------------------------
# 8. HANDLE MISSING NUMERIC VALUES
# ------------------------------------------------------------

numeric_cols = df.select_dtypes(
    include=np.number
).columns

df[numeric_cols] = df[numeric_cols].fillna(
    df[numeric_cols].median()
)

print("\nMissing Values After Imputation:")
print(df.isnull().sum())


# ------------------------------------------------------------
# 9. CHECK UNIQUE VALUES
# ------------------------------------------------------------

print("\nStatus Categories:")
print(df["Status"].unique())

print("\nNumber of Countries:")
print(df["Country"].nunique())

print("\nYears:")
print(df["Year"].unique())


# ------------------------------------------------------------
# 10. HIGH LIFE EXPECTANCY RECORDS
# ------------------------------------------------------------

high_life = df[
    df["Life expectancy"] > 75
]

print("\nHigh Life Expectancy Records:")
print(high_life.head())


# ------------------------------------------------------------
# 11. LOW LIFE EXPECTANCY RECORDS
# ------------------------------------------------------------

low_life = df[
    df["Life expectancy"] < 60
]

print("\nLow Life Expectancy Records:")
print(low_life.head())


# ------------------------------------------------------------
# 12. SORT BY LIFE EXPECTANCY
# ------------------------------------------------------------

sorted_df = df.sort_values(
    by="Life expectancy",
    ascending=False
)

print("\nTop Life Expectancy Records:")
print(sorted_df.head())


# ------------------------------------------------------------
# 13. MEAN
# ------------------------------------------------------------

print(
    "\nMean Life Expectancy :",
    df["Life expectancy"].mean()
)


# ------------------------------------------------------------
# 14. MEDIAN
# ------------------------------------------------------------

print(
    "Median Life Expectancy :",
    df["Life expectancy"].median()
)


# ------------------------------------------------------------
# 15. MODE
# ------------------------------------------------------------

print("\nMode Life Expectancy:")

print(
    df["Life expectancy"].mode()
)


# ------------------------------------------------------------
# 16. STANDARD DEVIATION
# ------------------------------------------------------------

print(
    "Standard Deviation :",
    df["Life expectancy"].std()
)


# ------------------------------------------------------------
# 17. AVERAGE LIFE EXPECTANCY BY STATUS
# ------------------------------------------------------------

status_avg = df.groupby(
    "Status"
)["Life expectancy"].mean()

print("\nAverage Life Expectancy by Status:")

print(
    status_avg.sort_values(
        ascending=False
    )
)


# ------------------------------------------------------------
# 18. AVERAGE LIFE EXPECTANCY BY YEAR
# ------------------------------------------------------------

year_avg = df.groupby(
    "Year"
)["Life expectancy"].mean()

print("\nAverage Life Expectancy by Year:")

print(year_avg)


# ------------------------------------------------------------
# 19. TOP 10 COUNTRIES
# ------------------------------------------------------------

country_avg = df.groupby(
    "Country"
)["Life expectancy"].mean()

top_10 = country_avg.sort_values(
    ascending=False
).head(10)

print("\nTop 10 Countries:")
print(top_10)


# ------------------------------------------------------------
# 20. BOTTOM 10 COUNTRIES
# ------------------------------------------------------------

bottom_10 = country_avg.sort_values(
    ascending=True
).head(10)

print("\nBottom 10 Countries:")
print(bottom_10)


# ============================================================
# VISUALIZATION
# ============================================================


# ------------------------------------------------------------
# 21. LIFE EXPECTANCY DISTRIBUTION
# ------------------------------------------------------------

plt.figure(figsize=(8,5))

sns.histplot(
    df["Life expectancy"],
    bins=30,
    kde=True
)

plt.title("Life Expectancy Distribution")

plt.xlabel("Life Expectancy")

plt.ylabel("Number of Records")

plt.show()


# ------------------------------------------------------------
# 22. LIFE EXPECTANCY OUTLIER DETECTION
# ------------------------------------------------------------

plt.figure(figsize=(8,5))

sns.boxplot(
    x=df["Life expectancy"]
)

plt.title("Life Expectancy Outlier Detection")

plt.xlabel("Life Expectancy")

plt.show()


# ------------------------------------------------------------
# 23. STATUS COUNT PLOT
# ------------------------------------------------------------

plt.figure(figsize=(7,5))

sns.countplot(
    data=df,
    x="Status"
)

plt.title("Country Status Distribution")

plt.xlabel("Country Status")

plt.ylabel("Number of Records")

plt.show()


# ------------------------------------------------------------
# 24. STATUS PIE CHART
# ------------------------------------------------------------

status = df["Status"].value_counts()

plt.figure(figsize=(8,8))

plt.pie(
    status,
    labels=status.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Developed vs Developing Countries")

plt.show()


# ------------------------------------------------------------
# 25. TOP 10 COUNTRIES BAR CHART
# ------------------------------------------------------------

plt.figure(figsize=(10,5))

top_10.plot(
    kind="bar"
)

plt.title(
    "Top 10 Countries by Life Expectancy"
)

plt.xlabel("Country")

plt.ylabel("Average Life Expectancy")

plt.xticks(rotation=45)

plt.show()


# ------------------------------------------------------------
# 26. BOTTOM 10 COUNTRIES BAR CHART
# ------------------------------------------------------------

plt.figure(figsize=(10,5))

bottom_10.plot(
    kind="bar"
)

plt.title(
    "Bottom 10 Countries by Life Expectancy"
)

plt.xlabel("Country")

plt.ylabel("Average Life Expectancy")

plt.xticks(rotation=45)

plt.show()


# ------------------------------------------------------------
# 27. LIFE EXPECTANCY OVER TIME
# ------------------------------------------------------------

plt.figure(figsize=(12,5))

plt.plot(
    year_avg.index,
    year_avg.values
)

plt.title(
    "Average Life Expectancy Over Time"
)

plt.xlabel("Year")

plt.ylabel("Average Life Expectancy")

plt.show()


# ------------------------------------------------------------
# 28. LIFE EXPECTANCY BY STATUS
# ------------------------------------------------------------

plt.figure(figsize=(8,5))

sns.boxplot(
    data=df,
    x="Status",
    y="Life expectancy"
)

plt.title(
    "Life Expectancy by Country Status"
)

plt.xlabel("Country Status")

plt.ylabel("Life Expectancy")

plt.show()


# ------------------------------------------------------------
# 29. VIOLIN PLOT
# ------------------------------------------------------------

plt.figure(figsize=(8,6))

sns.violinplot(
    data=df,
    x="Status",
    y="Life expectancy"
)

plt.title(
    "Life Expectancy Distribution by Status"
)

plt.xlabel("Country Status")

plt.ylabel("Life Expectancy")

plt.show()


# ------------------------------------------------------------
# 30. ADULT MORTALITY VS LIFE EXPECTANCY
# ------------------------------------------------------------

plt.figure(figsize=(8,5))

sns.scatterplot(
    data=df,
    x="Adult Mortality",
    y="Life expectancy"
)

plt.title(
    "Adult Mortality vs Life Expectancy"
)

plt.xlabel("Adult Mortality")

plt.ylabel("Life Expectancy")

plt.show()


# ------------------------------------------------------------
# 31. GDP VS LIFE EXPECTANCY
# ------------------------------------------------------------

plt.figure(figsize=(8,5))

sns.scatterplot(
    data=df,
    x="GDP",
    y="Life expectancy"
)

plt.title(
    "GDP vs Life Expectancy"
)

plt.xlabel("GDP")

plt.ylabel("Life Expectancy")

plt.show()


# ------------------------------------------------------------
# 32. SCHOOLING VS LIFE EXPECTANCY
# ------------------------------------------------------------

plt.figure(figsize=(8,5))

sns.scatterplot(
    data=df,
    x="Schooling",
    y="Life expectancy"
)

plt.title(
    "Schooling vs Life Expectancy"
)

plt.xlabel("Schooling")

plt.ylabel("Life Expectancy")

plt.show()


# ------------------------------------------------------------
# 33. BMI VS LIFE EXPECTANCY
# ------------------------------------------------------------

plt.figure(figsize=(8,5))

sns.scatterplot(
    data=df,
    x="BMI",
    y="Life expectancy"
)

plt.title(
    "BMI vs Life Expectancy"
)

plt.xlabel("BMI")

plt.ylabel("Life Expectancy")

plt.show()


# ------------------------------------------------------------
# 34. ALCOHOL VS LIFE EXPECTANCY
# ------------------------------------------------------------

plt.figure(figsize=(8,5))

sns.scatterplot(
    data=df,
    x="Alcohol",
    y="Life expectancy"
)

plt.title(
    "Alcohol Consumption vs Life Expectancy"
)

plt.xlabel("Alcohol Consumption")

plt.ylabel("Life Expectancy")

plt.show()


# ------------------------------------------------------------
# 35. POLIO VS LIFE EXPECTANCY
# ------------------------------------------------------------

plt.figure(figsize=(8,5))

sns.scatterplot(
    data=df,
    x="Polio",
    y="Life expectancy"
)

plt.title(
    "Polio Immunization vs Life Expectancy"
)

plt.xlabel("Polio Immunization")

plt.ylabel("Life Expectancy")

plt.show()


# ------------------------------------------------------------
# 36. DOLLAR EXPENDITURE VS LIFE EXPECTANCY
# ------------------------------------------------------------

plt.figure(figsize=(8,5))

sns.scatterplot(
    data=df,
    x="Total expenditure",
    y="Life expectancy"
)

plt.title(
    "Total Health Expenditure vs Life Expectancy"
)

plt.xlabel("Total Expenditure")

plt.ylabel("Life Expectancy")

plt.show()


# ------------------------------------------------------------
# 37. CORRELATION
# ------------------------------------------------------------

corr = df[
    [
        "Life expectancy",
        "Adult Mortality",
        "Alcohol",
        "BMI",
        "GDP",
        "Schooling",
        "Polio",
        "Total expenditure"
    ]
].corr()

print("\nCorrelation Matrix:")

print(corr)


# ------------------------------------------------------------
# 38. CORRELATION HEATMAP
# ------------------------------------------------------------

plt.figure(figsize=(10,7))

sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm"
)

plt.title(
    "Life Expectancy Correlation Heatmap"
)

plt.show()


# ------------------------------------------------------------
# 39. PAIR PLOT
# ------------------------------------------------------------

sns.pairplot(
    df[
        [
            "Life expectancy",
            "Adult Mortality",
            "Alcohol",
            "BMI",
            "GDP",
            "Schooling"
        ]
    ]
)

plt.show()


# ------------------------------------------------------------
# 40. BUBBLE CHART
# ------------------------------------------------------------

bubble_data = df.dropna(
    subset=[
        "GDP",
        "Life expectancy",
        "Population"
    ]
)

plt.figure(figsize=(9,6))

plt.scatter(
    bubble_data["GDP"],
    bubble_data["Life expectancy"],
    s=bubble_data["Population"] / 100000,
    alpha=0.5,
    c=bubble_data["Life expectancy"],
    cmap="viridis"
)

plt.colorbar(
    label="Life Expectancy"
)

plt.xlabel("GDP")

plt.ylabel("Life Expectancy")

plt.title(
    "Bubble Chart: GDP vs Life Expectancy"
)

plt.show()


# ------------------------------------------------------------
# 41. LIFE EXPECTANCY BY YEAR AND STATUS
# ------------------------------------------------------------

year_status = df.groupby(
    ["Year", "Status"]
)["Life expectancy"].mean().reset_index()

plt.figure(figsize=(12,6))

sns.lineplot(
    data=year_status,
    x="Year",
    y="Life expectancy",
    hue="Status",
    marker="o"
)

plt.title(
    "Life Expectancy Trend by Country Status"
)

plt.xlabel("Year")

plt.ylabel("Average Life Expectancy")

plt.show()


# ------------------------------------------------------------
# 42. TOP 10 LIFE EXPECTANCY RECORDS
# ------------------------------------------------------------

top_records = df.sort_values(
    by="Life expectancy",
    ascending=False
).head(10)

print("\nTop 10 Life Expectancy Records:")

print(
    top_records[
        [
            "Country",
            "Year",
            "Status",
            "Life expectancy",
            "GDP",
            "Schooling"
        ]
    ]
)


# ------------------------------------------------------------
# 43. LOWEST 10 LIFE EXPECTANCY RECORDS
# ------------------------------------------------------------

lowest_records = df.sort_values(
    by="Life expectancy",
    ascending=True
).head(10)

print("\nLowest 10 Life Expectancy Records:")

print(
    lowest_records[
        [
            "Country",
            "Year",
            "Status",
            "Life expectancy",
            "GDP",
            "Schooling"
        ]
    ]
)


# ============================================================
# PREPROCESSING
# ============================================================


# ------------------------------------------------------------
# 44. ENCODE STATUS
# ------------------------------------------------------------

encoder = LabelEncoder()

df["Status"] = encoder.fit_transform(
    df["Status"]
)

print("\nEncoded Status:")

print(df["Status"].head())


# ------------------------------------------------------------
# 45. ONE-HOT ENCODE COUNTRY
# ------------------------------------------------------------

country_encoded = pd.get_dummies(
    df["Country"]
)

print("\nCountry Encoded Data:")

print(country_encoded.head())


# ------------------------------------------------------------
# 46. NORMALIZE NUMERIC DATA
# ------------------------------------------------------------

numeric_cols = df.select_dtypes(
    include=np.number
).columns

scaler = MinMaxScaler()

df[numeric_cols] = scaler.fit_transform(
    df[numeric_cols]
)

print("\nNormalized Numeric Data:")

print(df.head())


# ------------------------------------------------------------
# 47. SAVE CLEANED DATASET
# ------------------------------------------------------------

df.to_csv(
    "cleaned_life_expectancy.csv",
    index=False
)

print(
    "\nCleaned dataset saved successfully."
)


# ------------------------------------------------------------
# 48. FINAL DATASET CHECK
# ------------------------------------------------------------

print(
    "\nFinal Dataset Shape :",
    df.shape
)

print("\nFinal Missing Values:")

print(df.isnull().sum())

print(
    "\nLife Expectancy Data Analysis Completed Successfully."
)
```
