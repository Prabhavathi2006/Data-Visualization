```python
# ============================================================
# LIFE EXPECTANCY DATA ANALYSIS
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns


# ------------------------------------------------------------
# 1. Load Dataset
# ------------------------------------------------------------

df = pd.read_csv("Life Expectancy Data.csv")

# Remove extra spaces from column names

df.columns = df.columns.str.strip()

print(df.head())

print(df.shape)

print(df.dtypes)


# ------------------------------------------------------------
# 2. Statistical Summary
# ------------------------------------------------------------

print(df.describe())


# ------------------------------------------------------------
# 3. Missing Values
# ------------------------------------------------------------

print("Missing Values\n")

print(df.isnull().sum())


# ------------------------------------------------------------
# 4. High Life Expectancy Countries
# ------------------------------------------------------------

high_life = df[
    df["Life expectancy"] >
    df["Life expectancy"].quantile(0.75)
]

print(high_life.head())


# ------------------------------------------------------------
# 5. Developed Countries
# ------------------------------------------------------------

developed = df[df["Status"] == "Developed"]

print(developed.head())


# ------------------------------------------------------------
# 6. Sort by Life Expectancy
# ------------------------------------------------------------

sorted_df = df.sort_values(
    by="Life expectancy",
    ascending=False
)

print(sorted_df.head())


# ------------------------------------------------------------
# 7. Mean, Median, Mode and Standard Deviation
# ------------------------------------------------------------

print("Mean :", df["Life expectancy"].mean())

print("Median :", df["Life expectancy"].median())

print("Mode :")

print(df["Life expectancy"].mode())

print(
    "Standard Deviation :",
    df["Life expectancy"].std()
)


# ------------------------------------------------------------
# 8. Average Life Expectancy by Status
# ------------------------------------------------------------

status_avg = df.groupby("Status")["Life expectancy"].mean()

print(
    status_avg.sort_values(
        ascending=False
    )
)


# ------------------------------------------------------------
# 9. Average Life Expectancy by Status - Bar Chart
# ------------------------------------------------------------

status_avg.sort_values(
    ascending=False
).plot(
    kind="bar"
)

plt.title("Average Life Expectancy by Country Status")

plt.xlabel("Country Status")

plt.ylabel("Average Life Expectancy")

plt.show()


# ------------------------------------------------------------
# 10. Average Life Expectancy by Year
# ------------------------------------------------------------

year_avg = df.groupby("Year")["Life expectancy"].mean()

print(year_avg)


# ------------------------------------------------------------
# 11. Life Expectancy Over Time
# ------------------------------------------------------------

plt.figure(figsize=(12,5))

plt.plot(
    year_avg.index,
    year_avg.values
)

plt.title("Average Life Expectancy Over Time")

plt.xlabel("Year")

plt.ylabel("Average Life Expectancy")

plt.show()


# ------------------------------------------------------------
# 12. Correlation Analysis
# ------------------------------------------------------------

corr = df[
    [
        "Life expectancy",
        "Adult Mortality",
        "Alcohol",
        "BMI",
        "GDP",
        "Schooling"
    ]
].corr()

print(corr)


# ------------------------------------------------------------
# 13. Correlation Heatmap
# ------------------------------------------------------------

plt.figure(figsize=(10,7))

sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm"
)

plt.title("Life Expectancy Correlation Heatmap")

plt.show()


# ------------------------------------------------------------
# 14. Adult Mortality vs Life Expectancy
# ------------------------------------------------------------

plt.figure(figsize=(8,5))

sns.scatterplot(
    data=df,
    x="Adult Mortality",
    y="Life expectancy"
)

plt.title("Adult Mortality vs Life Expectancy")

plt.xlabel("Adult Mortality")

plt.ylabel("Life Expectancy")

plt.show()


# ------------------------------------------------------------
# 15. GDP vs Life Expectancy
# ------------------------------------------------------------

plt.figure(figsize=(8,5))

sns.scatterplot(
    data=df,
    x="GDP",
    y="Life expectancy"
)

plt.title("GDP vs Life Expectancy")

plt.xlabel("GDP")

plt.ylabel("Life Expectancy")

plt.show()


# ------------------------------------------------------------
# 16. Schooling vs Life Expectancy
# ------------------------------------------------------------

plt.figure(figsize=(8,5))

sns.scatterplot(
    data=df,
    x="Schooling",
    y="Life expectancy"
)

plt.title("Schooling vs Life Expectancy")

plt.xlabel("Schooling")

plt.ylabel("Life Expectancy")

plt.show()


# ------------------------------------------------------------
# 17. BMI vs Life Expectancy
# ------------------------------------------------------------

plt.figure(figsize=(8,5))

sns.scatterplot(
    data=df,
    x="BMI",
    y="Life expectancy"
)

plt.title("BMI vs Life Expectancy")

plt.xlabel("BMI")

plt.ylabel("Life Expectancy")

plt.show()


# ------------------------------------------------------------
# 18. Life Expectancy Distribution
# ------------------------------------------------------------

plt.figure(figsize=(8,5))

sns.histplot(
    df["Life expectancy"],
    bins=30,
    kde=True
)

plt.title("Life Expectancy Distribution")

plt.xlabel("Life Expectancy")

plt.ylabel("Number of Countries")

plt.show()


# ------------------------------------------------------------
# 19. Life Expectancy Outlier Detection
# ------------------------------------------------------------

plt.figure(figsize=(8,5))

sns.boxplot(
    x=df["Life expectancy"]
)

plt.title("Life Expectancy Outlier Detection")

plt.show()


# ------------------------------------------------------------
# 20. Status Distribution
# ------------------------------------------------------------

plt.figure(figsize=(7,5))

sns.countplot(
    data=df,
    x="Status"
)

plt.title("Country Status Distribution")

plt.xlabel("Status")

plt.ylabel("Number of Records")

plt.show()


# ------------------------------------------------------------
# 21. Status Distribution Pie Chart
# ------------------------------------------------------------

status = df["Status"].value_counts()

plt.figure(figsize=(8,8))

plt.pie(
    status,
    labels=status.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Country Status Distribution")

plt.show()


# ------------------------------------------------------------
# 22. Top 10 Countries by Average Life Expectancy
# ------------------------------------------------------------

country_avg = df.groupby(
    "Country"
)["Life expectancy"].mean()

country_avg = country_avg.sort_values(
    ascending=False
).head(10)

print(country_avg)


# ------------------------------------------------------------
# 23. Top 10 Countries - Bar Chart
# ------------------------------------------------------------

plt.figure(figsize=(10,5))

country_avg.plot(
    kind="bar"
)

plt.title(
    "Top 10 Countries by Average Life Expectancy"
)

plt.xlabel("Country")

plt.ylabel("Average Life Expectancy")

plt.xticks(rotation=45)

plt.show()


# ------------------------------------------------------------
# 24. Bottom 10 Countries by Average Life Expectancy
# ------------------------------------------------------------

bottom_country = df.groupby(
    "Country"
)["Life expectancy"].mean()

bottom_country = bottom_country.sort_values(
    ascending=True
).head(10)

print(bottom_country)


# ------------------------------------------------------------
# 25. Bottom 10 Countries - Bar Chart
# ------------------------------------------------------------

plt.figure(figsize=(10,5))

bottom_country.plot(
    kind="bar"
)

plt.title(
    "Bottom 10 Countries by Average Life Expectancy"
)

plt.xlabel("Country")

plt.ylabel("Average Life Expectancy")

plt.xticks(rotation=45)

plt.show()


# ------------------------------------------------------------
# 26. Pair Plot
# ------------------------------------------------------------

sns.pairplot(
    df[
        [
            "Life expectancy",
            "Adult Mortality",
            "Alcohol",
            "BMI",
            "GDP",
            "Schooling"
        ]
    ]
)

plt.show()


# ------------------------------------------------------------
# 27. Bubble Chart
# GDP vs Life Expectancy
# ------------------------------------------------------------

bubble_data = df.dropna(
    subset=[
        "GDP",
        "Life expectancy",
        "Population"
    ]
)

plt.figure(figsize=(9,6))

plt.scatter(
    bubble_data["GDP"],
    bubble_data["Life expectancy"],
    s=bubble_data["Population"] / 100000,
    alpha=0.5,
    c=bubble_data["Life expectancy"],
    cmap="viridis"
)

plt.colorbar(
    label="Life Expectancy"
)

plt.xlabel("GDP")

plt.ylabel("Life Expectancy")

plt.title(
    "Bubble Chart: GDP vs Life Expectancy"
)

plt.show()


# ------------------------------------------------------------
# 28. Violin Plot
# Life Expectancy by Status
# ------------------------------------------------------------

plt.figure(figsize=(10,6))

sns.violinplot(
    data=df,
    x="Status",
    y="Life expectancy"
)

plt.title(
    "Life Expectancy Distribution by Country Status"
)

plt.xlabel("Country Status")

plt.ylabel("Life Expectancy")

plt.show()


# ------------------------------------------------------------
# 29. Box Plot
# Life Expectancy by Status
# ------------------------------------------------------------

plt.figure(figsize=(10,6))

sns.boxplot(
    data=df,
    x="Status",
    y="Life expectancy"
)

plt.title(
    "Life Expectancy by Country Status"
)

plt.xlabel("Country Status")

plt.ylabel("Life Expectancy")

plt.show()


# ------------------------------------------------------------
# 30. Life Expectancy vs Adult Mortality
# With Status
# ------------------------------------------------------------

plt.figure(figsize=(9,6))

sns.scatterplot(
    data=df,
    x="Adult Mortality",
    y="Life expectancy",
    hue="Status"
)

plt.title(
    "Adult Mortality vs Life Expectancy by Status"
)

plt.xlabel("Adult Mortality")

plt.ylabel("Life Expectancy")

plt.show()


# ------------------------------------------------------------
# 31. Life Expectancy vs Schooling
# With Status
# ------------------------------------------------------------

plt.figure(figsize=(9,6))

sns.scatterplot(
    data=df,
    x="Schooling",
    y="Life expectancy",
    hue="Status"
)

plt.title(
    "Schooling vs Life Expectancy by Status"
)

plt.xlabel("Schooling")

plt.ylabel("Life Expectancy")

plt.show()


# ------------------------------------------------------------
# 32. Average Life Expectancy by Year and Status
# ------------------------------------------------------------

year_status = df.groupby(
    ["Year", "Status"]
)["Life expectancy"].mean().reset_index()

plt.figure(figsize=(12,6))

sns.lineplot(
    data=year_status,
    x="Year",
    y="Life expectancy",
    hue="Status",
    marker="o"
)

plt.title(
    "Life Expectancy Trend by Country Status"
)

plt.xlabel("Year")

plt.ylabel("Average Life Expectancy")

plt.show()


# ------------------------------------------------------------
# 33. Top 10 Highest Life Expectancy Records
# ------------------------------------------------------------

top_10 = df.sort_values(
    by="Life expectancy",
    ascending=False
).head(10)

print(
    top_10[
        [
            "Country",
            "Year",
            "Status",
            "Life expectancy",
            "GDP",
            "Schooling"
        ]
    ]
)


# ------------------------------------------------------------
# 34. Lowest 10 Life Expectancy Records
# ------------------------------------------------------------

lowest_10 = df.sort_values(
    by="Life expectancy",
    ascending=True
).head(10)

print(
    lowest_10[
        [
            "Country",
            "Year",
            "Status",
            "Life expectancy",
            "GDP",
            "Schooling"
        ]
    ]
)


# ------------------------------------------------------------
# 35. Final Dataset Information
# ------------------------------------------------------------

print("Final Dataset Shape :", df.shape)

print("\nFinal Missing Values:")

print(df.isnull().sum())

print("\nLife Expectancy Data Analysis Completed Successfully.")
```
