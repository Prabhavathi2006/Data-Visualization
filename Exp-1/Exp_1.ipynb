```python
# ============================================================
# LIFE EXPECTANCY DATA ANALYSIS - COMPLETE CODE
# ============================================================

# -----------------------------
# 1. IMPORT LIBRARIES
# -----------------------------

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import MinMaxScaler, LabelEncoder


# -----------------------------
# 2. LOAD DATASET
# -----------------------------

df = pd.read_csv("Life Expectancy Data.csv")

# Remove unwanted spaces from column names
df.columns = df.columns.str.strip()

print("FIRST 5 ROWS")
print(df.head())

print("\nDATASET SHAPE")
print(df.shape)

print("\nDATA TYPES")
print(df.dtypes)


# -----------------------------
# 3. DATA INFORMATION
# -----------------------------

print("\nDATASET INFORMATION")
df.info()


# -----------------------------
# 4. STATISTICAL SUMMARY
# -----------------------------

print("\nSTATISTICAL SUMMARY")
print(df.describe())


# -----------------------------
# 5. MISSING VALUES
# -----------------------------

print("\nMISSING VALUES")
print(df.isnull().sum())


print("\nMISSING VALUE PERCENTAGE")
print((df.isnull().sum() / len(df)) * 100)


# -----------------------------
# 6. DUPLICATE VALUES
# -----------------------------

print("\nDUPLICATE ROWS")
print(df.duplicated().sum())

df.drop_duplicates(inplace=True)

print("SHAPE AFTER DUPLICATE REMOVAL")
print(df.shape)


# -----------------------------
# 7. HANDLE MISSING VALUES
# -----------------------------

numeric_cols = df.select_dtypes(
    include=np.number
).columns

df[numeric_cols] = df[numeric_cols].fillna(
    df[numeric_cols].median()
)

print("\nMISSING VALUES AFTER HANDLING")
print(df.isnull().sum())


# -----------------------------
# 8. UNIQUE VALUES
# -----------------------------

print("\nCOUNTRY STATUS")
print(df["Status"].unique())

print("\nNUMBER OF COUNTRIES")
print(df["Country"].nunique())

print("\nYEARS")
print(df["Year"].unique())


# -----------------------------
# 9. HIGH LIFE EXPECTANCY
# -----------------------------

high_life = df[
    df["Life expectancy"] > 75
]

print("\nHIGH LIFE EXPECTANCY RECORDS")
print(high_life.head())


# -----------------------------
# 10. LOW LIFE EXPECTANCY
# -----------------------------

low_life = df[
    df["Life expectancy"] < 60
]

print("\nLOW LIFE EXPECTANCY RECORDS")
print(low_life.head())


# -----------------------------
# 11. SORT DATA
# -----------------------------

sorted_df = df.sort_values(
    by="Life expectancy",
    ascending=False
)

print("\nTOP LIFE EXPECTANCY RECORDS")
print(sorted_df.head())


# -----------------------------
# 12. MEAN
# -----------------------------

print(
    "\nMEAN LIFE EXPECTANCY :",
    df["Life expectancy"].mean()
)


# -----------------------------
# 13. MEDIAN
# -----------------------------

print(
    "MEDIAN LIFE EXPECTANCY :",
    df["Life expectancy"].median()
)


# -----------------------------
# 14. MODE
# -----------------------------

print("\nMODE LIFE EXPECTANCY")

print(
    df["Life expectancy"].mode()
)


# -----------------------------
# 15. STANDARD DEVIATION
# -----------------------------

print(
    "STANDARD DEVIATION :",
    df["Life expectancy"].std()
)


# ============================================================
# GROUPING AND ANALYSIS
# ============================================================


# -----------------------------
# 16. AVERAGE BY STATUS
# -----------------------------

status_avg = df.groupby(
    "Status"
)["Life expectancy"].mean()

print("\nAVERAGE LIFE EXPECTANCY BY STATUS")

print(
    status_avg.sort_values(
        ascending=False
    )
)


# -----------------------------
# 17. AVERAGE BY YEAR
# -----------------------------

year_avg = df.groupby(
    "Year"
)["Life expectancy"].mean()

print("\nAVERAGE LIFE EXPECTANCY BY YEAR")

print(year_avg)


# -----------------------------
# 18. AVERAGE BY COUNTRY
# -----------------------------

country_avg = df.groupby(
    "Country"
)["Life expectancy"].mean()

print("\nTOP 10 COUNTRIES")

print(
    country_avg.sort_values(
        ascending=False
    ).head(10)
)


print("\nBOTTOM 10 COUNTRIES")

print(
    country_avg.sort_values(
        ascending=True
    ).head(10)
)


# ============================================================
# VISUALIZATION
# ============================================================


# -----------------------------
# 19. HISTOGRAM
# -----------------------------

plt.figure(figsize=(8,5))

sns.histplot(
    df["Life expectancy"],
    bins=30,
    kde=True
)

plt.title("Life Expectancy Distribution")
plt.xlabel("Life Expectancy")
plt.ylabel("Number of Records")

plt.show()


# -----------------------------
# 20. BOX PLOT
# -----------------------------

plt.figure(figsize=(8,5))

sns.boxplot(
    x=df["Life expectancy"]
)

plt.title("Life Expectancy Outlier Detection")
plt.xlabel("Life Expectancy")

plt.show()


# -----------------------------
# 21. STATUS COUNT PLOT
# -----------------------------

plt.figure(figsize=(7,5))

sns.countplot(
    data=df,
    x="Status"
)

plt.title("Country Status Distribution")
plt.xlabel("Status")
plt.ylabel("Number of Records")

plt.show()


# -----------------------------
# 22. STATUS PIE CHART
# -----------------------------

status = df["Status"].value_counts()

plt.figure(figsize=(8,8))

plt.pie(
    status,
    labels=status.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Developed vs Developing Countries")

plt.show()


# -----------------------------
# 23. TOP 10 COUNTRIES
# -----------------------------

top_10 = country_avg.sort_values(
    ascending=False
).head(10)

plt.figure(figsize=(10,5))

top_10.plot(
    kind="bar"
)

plt.title(
    "Top 10 Countries by Life Expectancy"
)

plt.xlabel("Country")
plt.ylabel("Average Life Expectancy")

plt.xticks(rotation=45)

plt.show()


# -----------------------------
# 24. BOTTOM 10 COUNTRIES
# -----------------------------

bottom_10 = country_avg.sort_values(
    ascending=True
).head(10)

plt.figure(figsize=(10,5))

bottom_10.plot(
    kind="bar"
)

plt.title(
    "Bottom 10 Countries by Life Expectancy"
)

plt.xlabel("Country")
plt.ylabel("Average Life Expectancy")

plt.xticks(rotation=45)

plt.show()


# -----------------------------
# 25. LIFE EXPECTANCY OVER TIME
# -----------------------------

plt.figure(figsize=(12,5))

plt.plot(
    year_avg.index,
    year_avg.values
)

plt.title(
    "Average Life Expectancy Over Time"
)

plt.xlabel("Year")
plt.ylabel("Average Life Expectancy")

plt.show()


# -----------------------------
# 26. LIFE EXPECTANCY BY STATUS
# -----------------------------

plt.figure(figsize=(8,5))

sns.boxplot(
    data=df,
    x="Status",
    y="Life expectancy"
)

plt.title(
    "Life Expectancy by Country Status"
)

plt.xlabel("Status")
plt.ylabel("Life Expectancy")

plt.show()


# -----------------------------
# 27. VIOLIN PLOT
# -----------------------------

plt.figure(figsize=(8,6))

sns.violinplot(
    data=df,
    x="Status",
    y="Life expectancy"
)

plt.title(
    "Life Expectancy Distribution by Status"
)

plt.xlabel("Status")
plt.ylabel("Life Expectancy")

plt.show()


# -----------------------------
# 28. ADULT MORTALITY VS LIFE EXPECTANCY
# -----------------------------

plt.figure(figsize=(8,5))

sns.scatterplot(
    data=df,
    x="Adult Mortality",
    y="Life expectancy"
)

plt.title(
    "Adult Mortality vs Life Expectancy"
)

plt.xlabel("Adult Mortality")
plt.ylabel("Life Expectancy")

plt.show()


# -----------------------------
# 29. GDP VS LIFE EXPECTANCY
# -----------------------------

plt.figure(figsize=(8,5))

sns.scatterplot(
    data=df,
    x="GDP",
    y="Life expectancy"
)

plt.title(
    "GDP vs Life Expectancy"
)

plt.xlabel("GDP")
plt.ylabel("Life Expectancy")

plt.show()


# -----------------------------
# 30. SCHOOLING VS LIFE EXPECTANCY
# -----------------------------

plt.figure(figsize=(8,5))

sns.scatterplot(
    data=df,
    x="Schooling",
    y="Life expectancy"
)

plt.title(
    "Schooling vs Life Expectancy"
)

plt.xlabel("Schooling")
plt.ylabel("Life Expectancy")

plt.show()


# -----------------------------
# 31. BMI VS LIFE EXPECTANCY
# -----------------------------

plt.figure(figsize=(8,5))

sns.scatterplot(
    data=df,
    x="BMI",
    y="Life expectancy"
)

plt.title(
    "BMI vs Life Expectancy"
)

plt.xlabel("BMI")
plt.ylabel("Life Expectancy")

plt.show()


# -----------------------------
# 32. ALCOHOL VS LIFE EXPECTANCY
# -----------------------------

plt.figure(figsize=(8,5))

sns.scatterplot(
    data=df,
    x="Alcohol",
    y="Life expectancy"
)

plt.title(
    "Alcohol Consumption vs Life Expectancy"
)

plt.xlabel("Alcohol")
plt.ylabel("Life Expectancy")

plt.show()


# -----------------------------
# 33. POLIO VS LIFE EXPECTANCY
# -----------------------------

plt.figure(figsize=(8,5))

sns.scatterplot(
    data=df,
    x="Polio",
    y="Life expectancy"
)

plt.title(
    "Polio Immunization vs Life Expectancy"
)

plt.xlabel("Polio Immunization")
plt.ylabel("Life Expectancy")

plt.show()


# -----------------------------
# 34. TOTAL EXPENDITURE VS LIFE EXPECTANCY
# -----------------------------

plt.figure(figsize=(8,5))

sns.scatterplot(
    data=df,
    x="Total expenditure",
    y="Life expectancy"
)

plt.title(
    "Total Health Expenditure vs Life Expectancy"
)

plt.xlabel("Total Expenditure")
plt.ylabel("Life Expectancy")

plt.show()


# ============================================================
# CORRELATION ANALYSIS
# ============================================================


# -----------------------------
# 35. CORRELATION MATRIX
# -----------------------------

corr = df[
    [
        "Life expectancy",
        "Adult Mortality",
        "Alcohol",
        "BMI",
        "GDP",
        "Schooling",
        "Polio",
        "Total expenditure"
    ]
].corr()

print("\nCORRELATION MATRIX")

print(corr)


# -----------------------------
# 36. CORRELATION HEATMAP
# -----------------------------

plt.figure(figsize=(10,7))

sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm"
)

plt.title(
    "Life Expectancy Correlation Heatmap"
)

plt.show()


# -----------------------------
# 37. PAIR PLOT
# -----------------------------

sns.pairplot(
    df[
        [
            "Life expectancy",
            "Adult Mortality",
            "Alcohol",
            "BMI",
            "GDP",
            "Schooling"
        ]
    ]
)

plt.show()


# -----------------------------
# 38. BUBBLE CHART
# -----------------------------

bubble_data = df.dropna(
    subset=[
        "GDP",
        "Life expectancy",
        "Population"
    ]
)

plt.figure(figsize=(9,6))

plt.scatter(
    bubble_data["GDP"],
    bubble_data["Life expectancy"],
    s=bubble_data["Population"] / 100000,
    alpha=0.5,
    c=bubble_data["Life expectancy"],
    cmap="viridis"
)

plt.colorbar(
    label="Life Expectancy"
)

plt.xlabel("GDP")
plt.ylabel("Life Expectancy")

plt.title(
    "Bubble Chart: GDP vs Life Expectancy"
)

plt.show()


# -----------------------------
# 39. YEAR AND STATUS ANALYSIS
# -----------------------------

year_status = df.groupby(
    ["Year", "Status"]
)["Life expectancy"].mean().reset_index()

plt.figure(figsize=(12,6))

sns.lineplot(
    data=year_status,
    x="Year",
    y="Life expectancy",
    hue="Status",
    marker="o"
)

plt.title(
    "Life Expectancy Trend by Country Status"
)

plt.xlabel("Year")
plt.ylabel("Average Life Expectancy")

plt.show()


# ============================================================
# TOP AND LOWEST RECORDS
# ============================================================


# -----------------------------
# 40. TOP 10 RECORDS
# -----------------------------

top_records = df.sort_values(
    by="Life expectancy",
    ascending=False
).head(10)

print("\nTOP 10 LIFE EXPECTANCY RECORDS")

print(
    top_records[
        [
            "Country",
            "Year",
            "Status",
            "Life expectancy",
            "GDP",
            "Schooling"
        ]
    ]
)


# -----------------------------
# 41. LOWEST 10 RECORDS
# -----------------------------

lowest_records = df.sort_values(
    by="Life expectancy",
    ascending=True
).head(10)

print("\nLOWEST 10 LIFE EXPECTANCY RECORDS")

print(
    lowest_records[
        [
            "Country",
            "Year",
            "Status",
            "Life expectancy",
            "GDP",
            "Schooling"
        ]
    ]
)


# ============================================================
# PREPROCESSING
# ============================================================


# -----------------------------
# 42. LABEL ENCODING
# -----------------------------

encoder = LabelEncoder()

df["Status"] = encoder.fit_transform(
    df["Status"]
)

print("\nENCODED STATUS")

print(df["Status"].head())


# -----------------------------
# 43. ONE-HOT ENCODING
# -----------------------------

country_encoded = pd.get_dummies(
    df["Country"]
)

print("\nONE-HOT ENCODED COUNTRY")

print(country_encoded.head())


# -----------------------------
# 44. NORMALIZATION
# -----------------------------

numeric_cols = df.select_dtypes(
    include=np.number
).columns

scaler = MinMaxScaler()

df[numeric_cols] = scaler.fit_transform(
    df[numeric_cols]
)

print("\nNORMALIZED DATA")

print(df.head())


# ============================================================
# SAVE CLEANED DATASET
# ============================================================


# -----------------------------
# 45. SAVE DATASET
# -----------------------------

df.to_csv(
    "cleaned_life_expectancy.csv",
    index=False
)

print(
    "\nCleaned dataset saved successfully."
)


# -----------------------------
# 46. FINAL CHECK
# -----------------------------

print(
    "\nFINAL DATASET SHAPE :",
    df.shape
)

print("\nFINAL MISSING VALUES")

print(df.isnull().sum())

print(
    "\nLIFE EXPECTANCY DATA ANALYSIS COMPLETED SUCCESSFULLY!"
)
```
